```python
from PIL import Image

# 1. Load ketiga gambar confusion matrix 
img1 = Image.open('cm_variety.png')   
img2 = Image.open('cm_ripeness.png') 
img3 = Image.open('cm_freshness.png')

# 2. Opsional: Samakan tinggi gambar jika berbeda 
height = img1.height
img2 = img2.resize((int(img2.width * height / img2.height), height))
img3 = img3.resize((int(img3.width * height / img3.height), height))

# 3. Hitung dimensi kanvas gabungan (lebar total, tinggi maksimum)
total_width = img1.width + img2.width + img3.width
max_height = height

# 4. Buat kanvas putih kosong
merged_image = Image.new('RGB', (total_width, max_height), color=(255, 255, 255))

# 5. Tempel (paste) gambar satu per satu dari kiri ke kanan
merged_image.paste(img1, (0, 0))
merged_image.paste(img2, (img1.width, 0))
merged_image.paste(img3, (img1.width + img2.width, 0))

# 6. Simpan hasilnya!
merged_image.save('confusion_matrices.png')
print("[INFO] Gambar berhasil digabungkan menjadi confusion_matrices.png!")
```
