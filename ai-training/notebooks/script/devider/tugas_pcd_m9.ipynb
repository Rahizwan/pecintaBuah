```python
from PIL import Image
import numpy as np
import pandas as pd
import os
from skimage.feature import graycomatrix, graycoprops

# Path dataset
dataset_path = 'part_4'
#saya dan tim tidak menggunakan seluruh citra dari dataset yang sudah kita sepakati, dikarenakan dataset yang kami gunakan sebelumnya untuk tugas besar berjumlah hampir 24 ribu gambar lebih.
# maka dari itu kami menggunakan beberapa saja dan tiap anggota kelompok mendatkan 50 citra unik untuk dianalisa di tugas ini

# Inisialisasi parameter GLCM
# Jarak d=1, Sudut 0, 45, 90, 135 derajat dalam radian
distances = [1]
angles = [0, np.pi/4, np.pi/2, 3*np.pi/4]

data_hasil = []

print(f"Memulai ekstraksi fitur dari folder: {dataset_path}")

# looping image 1-50
# saya menggunakan 50 images dan tiap anggota kelompok juga mendapatkan bagian nya 50 perorang
for i in range(1, 51):
    file_name = f"img_{i}.jpg"
    img_path = os.path.join(dataset_path, file_name)
    
    try:
        # Buka gambar, konversi ke Grayscale ('L'), lalu ubah ke NumPy Array
        img_raw = Image.open(img_path).convert('L')
        img = np.array(img_raw)
        
        # Hitung matriks GLCM
        glcm = graycomatrix(img, distances=distances, angles=angles, levels=256, 
                            symmetric=True, normed=True)
        
        # Ekstraksi fitur dan hitung rata-rata dari 4 sudut
        row = {
            'Citra': file_name,
            'kontras': np.mean(graycoprops(glcm, 'contrast')),
            'korelasi': np.mean(graycoprops(glcm, 'correlation')),
            'homogeneity': np.mean(graycoprops(glcm, 'homogeneity')),
            'energy': np.mean(graycoprops(glcm, 'energy'))
        }
        data_hasil.append(row)
        
    except FileNotFoundError:
        print(f"Peringatan: {file_name} tidak ditemukan di folder tersebut.")
    except Exception as e:
        print(f"Error pada {file_name}: {e}")

# membuat data frame
if data_hasil:
    df = pd.DataFrame(data_hasil)

    # tampilan terminal
    print("\n" + "="*65)
    print(f"{'HASIL EKSTRAKSI FITUR GLCM (RATA-RATA 4 SUDUT)':^65}")
    print("="*65)
    print(df.to_string(index=False)) 

    # export ke csv
    csv_filename = 'hasil_fitur_glcm_tugas.csv'
    df.to_csv(csv_filename, index=False)

    print("\n" + "="*65)
    print(f"Selesai! File disimpan sebagai: {csv_filename}")
    print("="*65)
else:
    print("Tidak ada data yang berhasil diproses. Cek kembali path folder Anda.")
```
